```python
%%writefile  A-Fundamentos.py

import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import firwin, lfilter, butter, freqz, hilbert

custom_css = """
<style>
    .stApp {
        background-color: #0e1a40;
        color: #E0E0E0;
        font-family: 'Courier New', monospace;
    }

    h1, h2, h3 {
        color: #00BFFF;
    }

    .main .block-container {
        padding-top: 2rem;
        padding-bottom: 2rem;
        background-color: #1B1D2B;
        border-radius: 10px;
    }

    section[data-testid="stSidebar"] {
        background-color: #222f5b;
        border-radius: 10px;
    }

    section[data-testid="stSidebar"] * {
        color: white !important;
    }

    .resultado-final {
        color: #FFD700;
        background-color: #2c3e50;
        border: 1px solid #FFD700;
        border-radius: 8px;
        padding: 1rem;
        text-align: center;
        font-size: 1.1rem;
    }

    .nota-info {
        color: #98FB98;
    }
</style>
"""

st.markdown(custom_css, unsafe_allow_html=True)

st.set_page_config(layout="wide", page_title="Procesamiento de Señales")

st.title("Procesamiento Digital de Señales")
st.markdown("""
Esta pestaña permite explorar conceptos fundamentales del procesamiento de señales.
Cada sección contiene una explicación teórica, un video de referencia y una simulación interactiva.
¡Usa los controles en la columna derecha para experimentar!
""")

video_urls = {
    "FT": "https://www.youtube.com/watch?v=spUNpyF58BY",
    "Filtros": "https://www.youtube.com/watch?v=Qmd685msY-Y",
    "Hilbert": "https://youtu.be/VyLU8hlhI-I?si=GzNoiYzA3_jxLGaJ",
    "QAM": "https://youtu.be/1asY7-NZ93g?si=aKIqMx7pf9ZuUU_R",
    "OFDM": "https://www.youtube.com/watch?v=VwnUuBshXGw"
}

# --- 1. Transformada de Fourier ---
with st.expander("1️⃣ Transformada de Fourier (FT, DFT, FFT)", expanded=True):
    col1, col2 = st.columns([1, 1])

    with col1:
        st.header("Análisis en Frecuencia")
        st.markdown(r"""
        La **Transformada de Fourier (FT)** permite representar una señal en el dominio de la frecuencia, descomponiéndola en una suma de senoidales. Es una herramienta central en el análisis de señales periódicas y aperiódicas.

        - **FT (Transformada de Fourier):** usada para señales continuas en tiempo continuo.
        - **DFT (Transformada Discreta de Fourier):** aplicada a secuencias finitas de datos.
        - **FFT (Fast Fourier Transform):** algoritmo eficiente para calcular la DFT.

        **Fórmulas clave:**

        - **FT:**
        $$ X(f) = \int_{-\infty}^{\infty} x(t)e^{-j2\pi ft}dt $$

        - **DFT:**
        $$ X[k] = \sum_{n=0}^{N-1} x[n] e^{-j \frac{2\pi}{N} kn} $$

        - **FFT:**
        Versión computacionalmente optimizada de la DFT, con complejidad \( O(N\log N) \).

        **Aplicaciones:**
        - Espectro de audio
        - Análisis de imágenes
        - Comunicaciones inalámbricas
        - Detección de patrones y filtrado
        """)
        st.video(video_urls["FT"])

    with col2:
        st.subheader("Simulación Interactiva")
        fs = st.slider("Frecuencia de Muestreo (Hz)", 100, 2000, 1000, key="fs_fft")
        f1 = st.slider("Frecuencia 1 (Hz)", 1, fs // 2 - 1, 50, key="f1_fft")
        f2 = st.slider("Frecuencia 2 (Hz)", 1, fs // 2 - 1, 120, key="f2_fft")

        t = np.arange(0, 1, 1/fs)
        x = np.sin(2*np.pi*f1*t) + 0.7*np.sin(2*np.pi*f2*t)

        X = np.fft.fft(x)
        f = np.fft.fftfreq(len(x), 1/fs)

        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6))
        plt.tight_layout(pad=3.0)
        ax1.plot(t, x)
        ax1.set_title("Señal en el Dominio del Tiempo")
        ax1.set_xlabel("Tiempo [s]")
        ax1.set_ylabel("Amplitud")
        ax1.grid(True)

        ax2.plot(f[:len(f)//2], np.abs(X)[:len(f)//2])
        ax2.set_title("Espectro de la Señal (FFT)")
        ax2.set_xlabel("Frecuencia [Hz]")
        ax2.set_ylabel("|X(f)|")
        ax2.grid(True)
        st.pyplot(fig)

# Resto del código se mantiene igual...


# --- 2. Filtrado Digital ---
with st.expander("2️⃣ Filtrado Digital (FIR/IIR)"):
    col1, col2 = st.columns([1, 1])

    with col1:
        st.header("Eliminación de Ruido")
        st.markdown(r"""
        Un **filtro digital** permite atenuar o eliminar frecuencias no deseadas (ruido) y dejar pasar las frecuencias de interés.
        - **FIR (Finite Impulse Response):** Siempre estables, pueden tener fase lineal.
        - **IIR (Infinite Impulse Response):** Más eficientes computacionalmente, pero la estabilidad no está garantizada.

        **Ecuación (FIR):**
        $$ y[n] = \sum_{k=0}^{M} b_k x[n-k] $$
        """)
        st.video(video_urls["Filtros"])

    with col2:
        st.subheader("Simulación Interactiva")
        fs_filt = 1000
        t_filt = np.arange(0, 1, 1/fs_filt)
        f_signal = st.slider("Frecuencia de la señal (Hz)", 1, 200, 50, key="f_signal_filt")
        f_noise = st.slider("Frecuencia del ruido (Hz)", 1, 500, 150, key="f_noise_filt")
        cutoff = st.slider("Frecuencia de Corte del Filtro (Hz)", 10, 400, 100, key="cutoff_filt")

        signal = np.sin(2*np.pi*f_signal*t_filt)
        noise = 0.5 * np.sin(2*np.pi*f_noise*t_filt)
        x_filt = signal + noise

        numtaps = 101 # Orden del filtro FIR
        b = firwin(numtaps, cutoff, fs=fs_filt, pass_zero='lowpass')
        y_filt = lfilter(b, [1.0], x_filt)

        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6))
        plt.tight_layout(pad=3.0)
        ax1.plot(t_filt, x_filt, label='Señal Original (con ruido)')
        ax1.plot(t_filt, y_filt, label='Señal Filtrada', linewidth=2)
        ax1.set_title("Señal Antes y Después del Filtrado")
        ax1.set_xlabel("Tiempo [s]")
        ax1.legend()
        ax1.grid(True)

        w, h = freqz(b, [1.0], fs=fs_filt)
        ax2.plot(w, 20 * np.log10(abs(h)))
        ax2.axvline(cutoff, color='red', linestyle='--', label=f'Corte ({cutoff} Hz)')
        ax2.set_title("Respuesta en Frecuencia del Filtro (Bode)")
        ax2.set_xlabel("Frecuencia [Hz]")
        ax2.set_ylabel("Ganancia [dB]")
        ax2.legend()
        ax2.grid(True)
        st.pyplot(fig)


# --- 3. Transformada de Hilbert ---
with st.expander("3️⃣ Señales Analíticas y Transformada de Hilbert"):
    col1, col2 = st.columns([1, 1])

    with col1:
        st.header("Señales en Fase y Cuadratura (I/Q)")
        st.markdown(r"""
        La **Transformada de Hilbert** genera una versión de la señal original desfasada en -90°. Al combinar la señal original (I) con su transformada (Q), se crea una **señal analítica**, fundamental para modulaciones y análisis de envolvente.

        **Señal Analítica:**
        $$ z(t) = x(t) + j \cdot \mathcal{H}\{x(t)\} $$
        Donde $\mathcal{H}\{x(t)\}$ es la transformada de Hilbert de $x(t)$.
        """)
        st.video(video_urls["Hilbert"])

    with col2:
        st.subheader("Simulación Interactiva")
        fs_hilbert = 1000
        t_hilbert = np.arange(0, 1, 1/fs_hilbert)
        f_hilbert = st.slider("Frecuencia de la señal (Hz)", 1, 200, 30, key="f_hilbert")
        amp_mod_freq = st.slider("Frecuencia de modulación de amplitud (Hz)", 0, 10, 5, key="amp_mod_freq")

        amplitude_envelope = 1 + 0.5 * np.sin(2 * np.pi * amp_mod_freq * t_hilbert)
        x_hilbert = amplitude_envelope * np.sin(2 * np.pi * f_hilbert * t_hilbert)
        analytic_signal = hilbert(x_hilbert)
        envelope_demod = np.abs(analytic_signal)

        fig, ax = plt.subplots(figsize=(8, 6))
        ax.plot(t_hilbert, x_hilbert, label='Señal Original Modulada')
        ax.plot(t_hilbert, envelope_demod, label='Envolvente Detectada (Hilbert)', linestyle='--', linewidth=2)
        ax.set_title("Detección de Envolvente con Hilbert")
        ax.set_xlabel("Tiempo [s]")
        ax.legend()
        ax.grid(True)
        st.pyplot(fig)


# --- 4. Modulación QAM ---
with st.expander("4️⃣ Señales I/Q y Modulación QAM"):
    col1, col2 = st.columns([1, 1])

    with col1:
        st.header("Transmisión Eficiente de Datos")
        st.markdown(r"""
        La **Modulación de Amplitud en Cuadratura (QAM)** usa dos portadoras (I y Q, desfasadas 90°) para transmitir múltiples bits por cada símbolo. Es la base de tecnologías modernas como Wi-Fi y 5G.

        **Ecuación de la señal:**
        $$ s(t) = I(t)\cos(2\pi f_c t) - Q(t)\sin(2\pi f_c t) $$
        """)
        st.video(video_urls["QAM"])

    with col2:
        st.subheader("Simulación Interactiva")
        M = st.select_slider("Orden de la Modulación (M-QAM)", options=[4, 16, 64, 256], value=16)
        num_symbols = st.slider("Número de Símbolos", 50, 1000, 200, key="num_symbols_qam")
        snr_db = st.slider("Relación Señal/Ruido (SNR) en dB", 0, 30, 25, key="snr_qam")

        # Generación de símbolos
        m_side = int(np.sqrt(M))
        bits = np.random.randint(0, M, num_symbols)
        I = 2 * (bits % m_side) - (m_side - 1)
        Q = 2 * (bits // m_side) - (m_side - 1)

        # Añadir ruido
        snr_linear = 10**(snr_db / 10)
        signal_power = np.mean(I**2 + Q**2)
        noise_power = signal_power / snr_linear
        noise = np.sqrt(noise_power / 2) * (np.random.randn(num_symbols) + 1j * np.random.randn(num_symbols))
        I_noisy = I + np.real(noise)
        Q_noisy = Q + np.imag(noise)

        fig, ax = plt.subplots(figsize=(6, 6))
        ax.scatter(I_noisy, Q_noisy, alpha=0.7, label=f'Símbolos con Ruido (SNR={snr_db}dB)')
        ax.scatter(I, Q, color='red', marker='x', s=100, label='Puntos de Constelación Ideales')
        ax.set_title(f"Diagrama de Constelación {M}-QAM")
        ax.set_xlabel("Componente en Fase (I)")
        ax.set_ylabel("Componente en Cuadratura (Q)")
        ax.grid(True)
        ax.axhline(0, color='black', lw=0.5)
        ax.axvline(0, color='black', lw=0.5)
        ax.set_aspect('equal', adjustable='box')
        ax.legend()
        st.pyplot(fig)

# --- 5. OFDM ---
with st.expander("5️⃣ OFDM (Orthogonal Frequency Division Multiplexing)"):
    col1, col2 = st.columns([1, 1])

    with col1:
        st.header("La Base del Wi-Fi y 5G")
        st.markdown("""
        **OFDM** es una técnica que divide un canal de comunicación de alta velocidad en cientos o miles de **subportadoras ortogonales** de baja velocidad.

        **Ventajas Clave:**
        - **Eficiencia Espectral:** Empaqueta las señales muy juntas sin interferencia.
        - **Robustez:** Muy resistente al desvanecimiento selectivo en frecuencia (multipath).
        - **Flexibilidad:** Cada subportadora puede usar una modulación diferente (como QAM).

        Es la tecnología que permite las altas velocidades de **Wi-Fi, 4G LTE y 5G**.
        """)
        st.video(video_urls["OFDM"])

    with col2:
        st.subheader("Visualización Conceptual")
        st.markdown("En esta visualización, puedes ver cómo las subportadoras de OFDM están espaciadas. El pico de cada portadora coincide con los nulos de las demás, lo que garantiza la **ortogonalidad** y evita la interferencia.")
        num_carriers = st.slider("Número de Subportadoras", 4, 16, 8, key="num_carriers_ofdm")

        fig, ax = plt.subplots(figsize=(8, 5))
        freq = np.linspace(-num_carriers, num_carriers, 1000)
        for i in range(-num_carriers//2, num_carriers//2):
            sinc = np.sinc(freq - i)**2
            ax.plot(freq, sinc, label=f'Subportadora {i+num_carriers//2+1}' if i < 2 else "_nolegend_")

        ax.set_title("Espectro de Subportadoras Ortogonales en OFDM")
        ax.set_xlabel("Frecuencia Normalizada")
        ax.set_ylabel("Densidad Espectral de Potencia")
        ax.set_xticks(np.arange(-num_carriers//2, num_carriers//2))
        ax.grid(True)
        st.pyplot(fig)
st.header("📚 Bibliografía y Recursos Recomendados")
st.markdown(r"""
A continuación se presentan algunas referencias clave para profundizar en los conceptos de **procesamiento de señales**, **modulación digital** y **tecnologías como OFDM y QAM**.

---

###  Libros recomendados

- **Oppenheim, A. V. & Schafer, R. W.**
  *Discrete-Time Signal Processing*. 3rd ed., Prentice Hall.
  Referencia esencial para transformadas, FFT y diseño de filtros digitales.

- **Proakis, J. G. & Manolakis, D. G.**
  *Digital Signal Processing: Principles, Algorithms, and Applications*. 4th ed., Pearson.
  Cobertura detallada de filtrado, análisis espectral y transformadas.

- **Haykin, S.**
  *Communication Systems*. 5th ed., Wiley.
  Clásico para comprender modulaciones I/Q, QAM y sistemas de comunicación.

- **Sklar, B.**
  *Digital Communications: Fundamentals and Applications*. 2nd ed., Prentice Hall.
  Excelente para OFDM, constelaciones y técnicas de modulación avanzadas.

- **Rappaport, T. S.**
  *Wireless Communications: Principles and Practice*. 2nd ed., Prentice Hall.
  Profundiza en aplicaciones modernas: LTE, Wi-Fi, 5G.

---

###  Recursos abiertos y prácticos

- **MIT OpenCourseWare – Signals and Systems**
  [ocw.mit.edu](https://ocw.mit.edu/courses/6-003-signals-and-systems-fall-2011/)
  Curso universitario gratuito con fundamentos de Fourier y filtrado.

- **Coursera – Digital Signal Processing (EPFL)**
  [Coursera DSP](https://www.coursera.org/learn/dsp)
  Curso online con ejercicios prácticos de filtrado, FFT y modulaciones.

- **YouTube – Wireless Future**
  [Wireless Future](https://www.youtube.com/@WirelessFuture)
  Canal con explicaciones de QAM, OFDM y tecnologías inalámbricas.

- **Documentación SciPy Signal**
  [SciPy Signal Processing](https://docs.scipy.org/doc/scipy/reference/signal.html)
  Funciones `fft`, `hilbert`, `butter`, `firwin` para procesamiento digital en Python.

- **GNU Radio Project**
  [gnuradio.org](https://www.gnuradio.org/)
  Herramienta de radio definida por software (SDR) para practicar señales I/Q y OFDM.

---

**Tip:** Estos materiales complementan la parte práctica de tu dashboard interactivo. Se recomienda explorar los enlaces y libros según tu nivel de profundidad.
""")
with st.sidebar.expander("👥 Integrantes del Proyecto"):
    st.markdown("""
    ### Equipo de Trabajo

    Este proyecto fue desarrollado por estudiantes de la **Universidad Nacional de Colombia**, como parte del curso de **de Señales y Sistemas**.

    **Integrantes:**
    - 📌 **Sofía López Usma**
    - 📌 **Miguel Andrés Peña**
    - 📌 **Juan Fernando Montoya Ortiz**


    """)

```

```python
%%writefile  B-Dominio_de_Frecuencia.py
# app.py
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq
from scipy.signal import butter, filtfilt, freqs

custom_css = """
<style>
    .stApp {
        background-color: #0e1a40;
        color: #E0E0E0;
        font-family: 'Courier New', monospace;
    }

    h1, h2, h3 {
        color: #00BFFF;
    }

    .main .block-container {
        padding-top: 2rem;
        padding-bottom: 2rem;
        background-color: #1B1D2B;
        border-radius: 10px;
    }

    section[data-testid="stSidebar"] {
        background-color: #222f5b;
        border-radius: 10px;
    }

    section[data-testid="stSidebar"] * {
        color: white !important;
    }

    .resultado-final {
        color: #FFD700;
        background-color: #2c3e50;
        border: 1px solid #FFD700;
        border-radius: 8px;
        padding: 1rem;
        text-align: center;
        font-size: 1.1rem;
    }

    .nota-info {
        color: #98FB98;
    }
</style>
"""

st.markdown(custom_css, unsafe_allow_html=True)

st.set_page_config(layout="wide", page_title="Procesamiento de Señales")


st.title("🔍 Análisis en el Dominio de la Frecuencia")

# =========================================
# 🔧 Controles interactivos (sidebar)
# =========================================
st.sidebar.header("Controles de la señal")
amp1 = st.sidebar.slider("Amplitud de 50 Hz", 0.0, 2.0, 1.0, step=0.1)
amp2 = st.sidebar.slider("Amplitud de 150 Hz", 0.0, 2.0, 0.5, step=0.1)
noise_level = st.sidebar.slider("Nivel de ruido aleatorio", 0.0, 1.0, 0.3, step=0.05)
fc = st.sidebar.slider("Frecuencia de corte del filtro (Hz)", 10, 500, 100, step=10)

# =========================================
# 1️⃣ Generar señal sintética
# =========================================
st.header("1. Generar señal sintética")
fs = 1000  # Frecuencia de muestreo (Hz)
t = np.linspace(0, 1, fs, endpoint=False)  # 1 segundo
x = amp1 * np.sin(2 * np.pi * 50 * t) + amp2 * np.sin(2 * np.pi * 150 * t) + noise_level * np.random.randn(len(t))

fig1, ax1 = plt.subplots(figsize=(10, 3))
ax1.plot(t, x)
ax1.set_title("Señal sintética en el tiempo")
ax1.set_xlabel("Tiempo (s)")
ax1.set_ylabel("Amplitud")
ax1.grid()
st.pyplot(fig1)

# =========================================
# 2️⃣ Aplicar FFT y visualizar en frecuencia
# =========================================
st.header("2. Transformada Rápida de Fourier (FFT)")

X = fft(x)
freq = fftfreq(len(t), 1/fs)
X_mag = np.abs(X[:len(t)//2])
freq_pos = freq[:len(t)//2]

fig2, ax2 = plt.subplots(figsize=(10, 3))
ax2.plot(freq_pos, X_mag)
ax2.set_title("FFT - Magnitud de la señal")
ax2.set_xlabel("Frecuencia (Hz)")
ax2.set_ylabel("Magnitud")
ax2.grid()
st.pyplot(fig2)

# =========================================
# 3️⃣ Filtro pasa-bajo
# =========================================
st.header("3. Aplicar filtro pasa-bajo (Butterworth)")

b, a = butter(N=4, Wn=fc / (fs / 2), btype='low')
x_filtered = filtfilt(b, a, x)

fig3, ax3 = plt.subplots(figsize=(10, 3))
ax3.plot(t, x, label="Original")
ax3.plot(t, x_filtered, label="Filtrada", linewidth=2)
ax3.set_title("Señal antes y después del filtro pasa-bajo")
ax3.set_xlabel("Tiempo (s)")
ax3.legend()
ax3.grid()
st.pyplot(fig3)

# =========================================
# 4️⃣ FFT de la señal filtrada
# =========================================
st.header("4. FFT de la señal filtrada")

Xf = fft(x_filtered)
Xf_mag = np.abs(Xf[:len(t)//2])

fig4, ax4 = plt.subplots(figsize=(10, 3))
ax4.plot(freq_pos, Xf_mag)
ax4.set_title("FFT - Magnitud de la señal filtrada")
ax4.set_xlabel("Frecuencia (Hz)")
ax4.set_ylabel("Magnitud")
ax4.grid()
st.pyplot(fig4)

# =========================================
# 5️⃣ Diagrama de Bode
# =========================================
st.header("5. Diagrama de Bode del filtro")

w, h = freqs(b, a, worN=8000)
frequencies_hz = w * fs / (2 * np.pi)

fig5, (ax5a, ax5b) = plt.subplots(2, 1, figsize=(10, 5))

# Magnitud
ax5a.semilogx(frequencies_hz, 20 * np.log10(abs(h)))
ax5a.set_title('Bode: Magnitud del filtro pasa-bajo')
ax5a.set_ylabel('Magnitud (dB)')
ax5a.grid(which='both', axis='both')

# Fase
ax5b.semilogx(frequencies_hz, np.angle(h))
ax5b.set_xlabel('Frecuencia (Hz)')
ax5b.set_ylabel('Fase (radianes)')
ax5b.grid(which='both', axis='both')

st.pyplot(fig5)

with st.sidebar.expander("👥 Integrantes del Proyecto"):
    st.markdown("""
    ### Equipo de Trabajo

    Este proyecto fue desarrollado por estudiantes de la **Universidad Nacional de Colombia**, como parte del curso de **de Señales y Sistemas**.

    **Integrantes:**
    - 📌 **Sofía López Usma**
    - 📌 **Miguel Andrés Peña**
    - 📌 **Juan Fernando Montoya Ortiz**


    """)
```

```python
%%writefile C-Construyendo_las_Señales.py

import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import hilbert

custom_css = """
<style>
    .stApp {
        background-color: #0e1a40;
        color: #E0E0E0;
        font-family: 'Courier New', monospace;
    }

    h1, h2, h3 {
        color: #00BFFF;
    }

    .main .block-container {
        padding-top: 2rem;
        padding-bottom: 2rem;
        background-color: #1B1D2B;
        border-radius: 10px;
    }

    section[data-testid="stSidebar"] {
        background-color: #222f5b;
        border-radius: 10px;
    }

    section[data-testid="stSidebar"] * {
        color: white !important;
    }

    .resultado-final {
        color: #FFD700;
        background-color: #2c3e50;
        border: 1px solid #FFD700;
        border-radius: 8px;
        padding: 1rem;
        text-align: center;
        font-size: 1.1rem;
    }

    .nota-info {
        color: #98FB98;
    }
</style>
"""

st.markdown(custom_css, unsafe_allow_html=True)

st.set_page_config(layout="wide", page_title="Procesamiento de Señales")


st.title("Construyendo las Señales I/Q usando la Transformada de Hilbert")



st.markdown("""


Este notebook muestra cómo construir señales en cuadratura (I/Q) a partir de una señal real. Utilizaremos la **Transformada de Hilbert** para obtener la componente Q (cuadratura), mientras que la señal original será la componente I (en fase).

Este método se usa en procesamiento de señales y telecomunicaciones, especialmente en modulaciones como QAM y SSB.

##  Fundamento Teórico

### Señales I/Q (In-Phase y Quadrature)

Las señales **I/Q** (en fase y en cuadratura) son una representación fundamental en procesamiento de señales, comunicaciones digitales, modulación y demodulación. Esta representación permite expresar cualquier señal de banda limitada como una combinación de dos señales ortogonales (es decir, con 90° de desfase entre sí).

Una señal I/Q tiene la siguiente forma general:
""")
st.latex(r"s(t) = I(t) + jQ(t)")

st.markdown("""
donde:
- \( I(t) \): Componente **en fase**, usualmente una portadora cosenoidal.
- \( Q(t) \): Componente **en cuadratura**, obtenida como una copia de \( I(t) \) pero con un desfase de 90° (π/2 radianes).
- \( j \): Unidad imaginaria.

Esta forma es conocida como una **señal analítica**, y tiene ventajas clave:
- Permite separar la **información de amplitud y fase** de una señal.
- Facilita la modulación en cuadratura como QAM (Quadrature Amplitude Modulation).
- Se usa en demodulación coherente, procesamiento digital de señales, SDR (Software Defined Radio), etc.

#### Ejemplo práctico:
""")
st.markdown("""Si tomamos:""")
st.latex(r"I(t) = \cos(2\pi f t)")
st.markdown("""Entonces""")
st.latex(r"Q(t) = \sin(2\pi f t) = \cos(2\pi f t - \\frac{\pi}{2})")
st.markdown("""Esto representa una señal rotando en el plano complejo como:""")
st.latex(r"s(t) = \cos(2\pi f t) + j\sin(2\pi f t) = e^{j2\pi f t}")

st.markdown("""
---

### ⚙️ Transformada de Hilbert

La **Transformada de Hilbert** es una herramienta del análisis de señales que permite obtener la señal en cuadratura a partir de una señal real.

Dada una señal real \( x(t) \), la transformada de Hilbert \( \hat{x}(t) \) se define como:
""")
st.latex(r"\hat{x}(t) = \mathcal{H}\{x(t)\} = \\frac{1}{\pi} \, P \int_{-\infty}^{\infty} \\frac{x(	au)}{t - 	au} d	au")

st.markdown("""
- \( \mathcal{H}\{x(t)\} \): Transformada de Hilbert de \( x(t) \)
- \( P \): símbolo de la integral principal de Cauchy (para tratar la singularidad en \( t = 	au \))

La señal **analítica** resultante es:
""")
st.latex(r"x_a(t) = x(t) + j\hat{x}(t)")

st.markdown("""
- Parte real: \( x(t) \) → componente I.
- Parte imaginaria: \( \hat{x}(t) \) → componente Q.

La función \( \hat{x}(t) \) es la versión en cuadratura de la señal original. Esta operación:
- Elimina todas las componentes negativas de frecuencia.
- Conserva solo las positivas, con un desfase de **+90°**.
- Equivale a aplicar un **filtro pasa banda** centrado en frecuencia positiva.

###  Propiedades de la Transformada de Hilbert

1. **Fase**: Desplaza las componentes positivas en +90° y las negativas en −90°.
2. **Magnitud**: No altera la magnitud de la señal original.
3. **Ortogonalidad**: \( x(t) \) y \( \hat{x}(t) \) son ortogonales en el sentido de energía.

###  Aplicaciones de señales I/Q

- **Modulación digital** (QAM, PSK)
- **Procesamiento de señales de radar**
- **Software Defined Radio (SDR)**
- **Demodulación en comunicaciones inalámbricas**
- **Análisis espectral y filtrado adaptativo**

En este notebook implementaremos este modelo generando una señal \( I(t) = \cos(2\pi f t) \), aplicaremos la transformada de Hilbert para obtener \( Q(t) \), y analizaremos ambas en el dominio del tiempo y la frecuencia.

###  Recurso recomendado: Video educativo sobre señales I/Q

Para complementar la teoría con una explicación visual clara y accesible, se recomienda ver el siguiente video:

*El siguiente video te ayudará a entender mejor la Transformada de Hilbert.*
 **Canal**: Universitat Politècnica de València
 **Duración**: ~10 minutos

""")
st.video("https://www.youtube.com/watch?v=OBUxt1suwLw&t=1s")

st.markdown("""
#### ¿Qué aprenderás en el video?

- Qué representan las señales I (In-phase) y Q (Quadrature).
- Cómo se combinan para representar una señal compleja.
- Qué ocurre en el plano I/Q (plano complejo o plano de constelación).
""")

st.sidebar.header("Configuración de la señal")
fs = st.sidebar.slider("Frecuencia de muestreo [Hz]", 500, 5000, 1000)
f = st.sidebar.slider("Frecuencia de la señal [Hz]", 10, 500, 50)
duracion = st.sidebar.slider("Duración [s]", 0.1, 10.0, 2.5, step=0.1)
tipo = st.sidebar.selectbox("Tipo de señal", ["Coseno", "Seno"])

t = np.linspace(0, duracion, int(fs * duracion), endpoint=False)
if tipo == "Coseno":
    x = np.cos(2 * np.pi * f * t)
elif tipo == "Seno":
    x = np.sin(2 * np.pi * f * t)

xh = hilbert(x)
I = x
Q = np.imag(xh)

st.subheader(" Señales I y Q en el dominio del tiempo")
fig1, ax1 = plt.subplots()
ax1.plot(t, I, label="I (In-phase)")
ax1.plot(t, Q, label="Q (Quadrature)", linestyle='--')
ax1.set_xlabel("Tiempo [s]")
ax1.set_ylabel("Amplitud")
ax1.legend()
ax1.grid()
st.pyplot(fig1)

st.markdown("""

En la primera gráfica observamos las señales \( I(t) \) y \( Q(t) \):

- La señal **I(t)** (en azul) es un coseno de frecuencia fija, correspondiente a la componente **en fase**.
- La señal **Q(t)** (en naranja o línea punteada) es su transformada de Hilbert: un seno con la misma frecuencia pero **desfasado 90°** respecto a I(t). Esta componente representa la **cuadratura**.

 **Observaciones**:
- Ambas señales tienen igual amplitud y frecuencia.
- La diferencia de fase constante de 90° confirma que Q es ortogonal a I.
- La ortogonalidad es esencial para construir una señal compleja (analítica) que puede portar más información en modulaciones como QAM o PSK.""")


X = np.fft.fft(xh)
freq = np.fft.fftfreq(len(X), d=1/fs)

st.subheader(" Espectro de Magnitud (dB)")
fig2, ax2 = plt.subplots()
ax2.plot(freq[:len(freq)//2], 20 * np.log10(np.abs(X[:len(freq)//2])))
ax2.set_xlabel("Frecuencia [Hz]")
ax2.set_ylabel("Magnitud [dB]")
ax2.grid()
st.pyplot(fig2)

st.markdown("""

La primera mitad del espectro muestra la **magnitud** en decibelios (dB) de la señal analítica \( s(t) = I(t) + jQ(t) \):

- El pico principal está centrado en la frecuencia de la señal (por ejemplo, 50 Hz), como se espera de una señal monofrecuencia.
- La transformada de Hilbert **suprime** las componentes negativas de frecuencia, lo cual es consistente con la definición de una señal **analítica**.

 **Observaciones**:
- La energía de la señal está concentrada únicamente en las frecuencias positivas.
- El espectro tiene una sola banda, lo que facilita el procesamiento y la modulación.""")


st.subheader(" Espectro de Fase")
fig3, ax3 = plt.subplots()
ax3.plot(freq[:len(freq)//2], np.angle(X[:len(freq)//2]))
ax3.set_xlabel("Frecuencia [Hz]")
ax3.set_ylabel("Fase [rad]")
ax3.grid()
st.pyplot(fig3)

st.markdown("""
La segunda gráfica del dominio de la frecuencia representa la **fase** de la señal analítica:

- La fase cambia suavemente cerca de la frecuencia dominante (50 Hz).
- Dado que la transformada de Hilbert altera solo la fase (no la magnitud), la gráfica de fase contiene la información más importante sobre el **desplazamiento angular** entre I y Q.

 **Observaciones**:
- Se puede observar una transición en la fase en la banda donde existe energía (alrededor de 50 Hz).
- El valor de fase tiende a 90° en esa banda, validando el comportamiento ideal de la transformada de Hilbert.""")

with st.sidebar.expander("👥 Integrantes del Proyecto"):
    st.markdown("""
    ### Equipo de Trabajo

    Este proyecto fue desarrollado por estudiantes de la **Universidad Nacional de Colombia**, como parte del curso de **de Señales y Sistemas**.

    **Integrantes:**
    - 📌 **Sofía López Usma**
    - 📌 **Miguel Andrés Peña**
    - 📌 **Juan Fernando Montoya Ortiz**


    """)
```

```python
%%writefile D-Modulacion_QAM.py

import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq, fftshift



custom_css = """
<style>
    .stApp {
        background-color: #0e1a40;
        color: #E0E0E0;
        font-family: 'Courier New', monospace;
    }

    h1, h2, h3 {
        color: #00BFFF;
    }

    .main .block-container {
        padding-top: 2rem;
        padding-bottom: 2rem;
        background-color: #1B1D2B;
        border-radius: 10px;
    }

    section[data-testid="stSidebar"] {
        background-color: #222f5b;
        border-radius: 10px;
    }

    section[data-testid="stSidebar"] * {
        color: white !important;
    }

    .resultado-final {
        color: #FFD700;
        background-color: #2c3e50;
        border: 1px solid #FFD700;
        border-radius: 8px;
        padding: 1rem;
        text-align: center;
        font-size: 1.1rem;
    }

    .nota-info {
        color: #98FB98;
    }
</style>
"""

st.markdown(custom_css, unsafe_allow_html=True)

st.set_page_config(layout="wide", page_title="Procesamiento de Señales")



# --- Configuración de la Página ---
st.set_page_config(
    layout="wide",
    page_title="Simulador Avanzado de Modulación QAM",
    page_icon="📡"
)

plt.style.use('seaborn-v0_8-darkgrid')


# --- FUNCIONES ---

def generar_bits(num_simbolos, k):
    return np.random.randint(0, 2, num_simbolos * k)


def mapear_bits_a_simbolos_qam(bits, M, k):
    if k % 2 != 0:
        raise ValueError("k debe ser par para QAM cuadrado.")

    simbolos_decimales = bits.reshape(-1, k).dot(1 << np.arange(k - 1, -1, -1))
    lado = int(np.sqrt(M))
    amplitudes = np.arange(-(lado - 1), lado, 2)

    simbolos_I = amplitudes[simbolos_decimales % lado]
    simbolos_Q = amplitudes[simbolos_decimales // lado]

    energia_media = np.mean(simbolos_I ** 2 + simbolos_Q ** 2)
    escala = np.sqrt(energia_media)

    simbolos_I = simbolos_I / escala
    simbolos_Q = simbolos_Q / escala

    return simbolos_I, simbolos_Q, escala


def modular_qam(simbolos_I, simbolos_Q, fc, fs, tasa_simbolo):
    muestras_por_simbolo = int(fs / tasa_simbolo)
    I_t = np.repeat(simbolos_I, muestras_por_simbolo)
    Q_t = np.repeat(simbolos_Q, muestras_por_simbolo)

    t = np.linspace(0, len(I_t) / fs, len(I_t), endpoint=False)

    portadora_I = np.cos(2 * np.pi * fc * t)
    portadora_Q = np.sin(2 * np.pi * fc * t)

    s_t = I_t * portadora_I - Q_t * portadora_Q

    return s_t, t


def canal_awgn(signal, snr_db):
    potencia_signal = np.mean(signal ** 2)
    snr_lineal = 10 ** (snr_db / 10)
    potencia_ruido = potencia_signal / snr_lineal

    ruido = np.sqrt(potencia_ruido) * np.random.randn(len(signal))
    return signal + ruido


def demodular_y_decidir(signal_recibida, fc, fs, tasa_simbolo, M, k, escala):
    muestras_por_simbolo = int(fs / tasa_simbolo)
    num_simbolos = len(signal_recibida) // muestras_por_simbolo

    t = np.linspace(0, len(signal_recibida) / fs, len(signal_recibida), endpoint=False)

    demod_I = signal_recibida * np.cos(2 * np.pi * fc * t) * 2
    demod_Q = signal_recibida * -np.sin(2 * np.pi * fc * t) * 2

    simbolos_I_rec = np.mean(demod_I.reshape(num_simbolos, muestras_por_simbolo), axis=1) / escala
    simbolos_Q_rec = np.mean(demod_Q.reshape(num_simbolos, muestras_por_simbolo), axis=1) / escala

    lado = int(np.sqrt(M))
    amplitudes = np.arange(-(lado - 1), lado, 2)

    indices_I = np.argmin(np.abs(simbolos_I_rec[:, None] - amplitudes[None, :]), axis=1)
    indices_Q = np.argmin(np.abs(simbolos_Q_rec[:, None] - amplitudes[None, :]), axis=1)

    simbolos_dec = indices_Q * lado + indices_I

    bits_rx = ((simbolos_dec[:, None] & (1 << np.arange(k - 1, -1, -1))) > 0).astype(int).flatten()

    return simbolos_I_rec, simbolos_Q_rec, bits_rx


def calcular_ber(bits_tx, bits_rx):
    N = min(len(bits_tx), len(bits_rx))
    errores = np.sum(bits_tx[:N] != bits_rx[:N])
    return errores, N, errores / N if N > 0 else 0


# --- UI ---

st.title(" Simulador Avanzado de Transmisión Digital QAM")

st.markdown(r"""
Este simulador muestra cómo funciona una cadena de transmisión **QAM (Quadrature Amplitude Modulation)** con canal ruidoso (AWGN).

**Ecuación de la señal:**

$$ s(t) = I(t)\cos(2\pi f_c t) - Q(t)\sin(2\pi f_c t) $$

**Canal:**

$$ r(t) = s(t) + n(t) $$
""")

st.header(" Teoría de la Modulación 16-QAM")

st.markdown(r"""
##  ¿Qué es QAM?

**QAM (Quadrature Amplitude Modulation)** combina modulación en amplitud y fase. Usa dos portadoras ortogonales:
- Una en fase (*In-phase*, I)
- Otra en cuadratura (*Quadrature*, Q)

Cada portadora lleva parte de la información y se combinan para transmitir más bits por símbolo.

---

##  16-QAM

En **16-QAM**, cada símbolo codifica **4 bits**:
- 2 bits definen el nivel de la señal I.
- 2 bits definen el nivel de la señal Q.

Esto genera 16 puntos posibles (una constelación **4×4**) en el plano I-Q.

---

##  Mapeo de Símbolos

Cada grupo de 4 bits se asigna a un par (I, Q):

| Bits | I | Q |
|------|----|----|
| 0000 | -3 | -3 |
| 0001 | -3 | -1 |
| 0010 | -3 | +3 |
| ...  | ... | ... |
| 1111 | +3 | +3 |

Los niveles se normalizan para que la potencia promedio sea constante.

---

##  Señales I(t) y Q(t)

La señal modulada se expresa como:

$$ s(t) = I(t)\cos(2\pi f_c t) - Q(t)\sin(2\pi f_c t) $$

Donde:
- \( I(t) \) es la componente en fase.
- \( Q(t) \) es la componente en cuadratura.
- \( f_c \) es la frecuencia de la portadora.

Durante cada símbolo, \( I(t) \) y \( Q(t) \) son constantes.

---

##  Modulación sobre la Portadora

**Proceso:**
1. Generar bits.
2. Mapearlos a \( (I, Q) \).
3. Multiplicar \( I \) por una portadora coseno.
4. Multiplicar \( Q \) por una portadora seno (desfasada 90°).
5. Combinar para formar \( s(t) \).

---

## Espectro

El espectro de QAM muestra dos lóbulos centrados en la frecuencia de portadora \( f_c \). Su ancho depende de la tasa de símbolos.

---

##  Diagrama de Constelación

Representa todos los puntos posibles \( (I, Q) \).
- Cada punto es un símbolo.
- La separación entre puntos determina la robustez al ruido.
- Desviaciones por interferencia distorsionan la constelación.

---

##  Resumen

| Concepto | Descripción |
|----------------|----------------|
| **QAM** | Modula amplitud y fase |
| **16-QAM** | 16 símbolos (4 bits por símbolo) |
| **I(t)** | Componente en fase |
| **Q(t)** | Componente en cuadratura |
| **Señal modulada** | \( s(t) = I(t)\cos(2\pi f_c t) - Q(t)\sin(2\pi f_c t) \) |
| **Constelación** | Plano I-Q |
| **Espectro** | Banda centrada en \( f_c \) |

""")


# --- Sidebar ---
st.sidebar.header(" Parámetros")
M = st.sidebar.selectbox("Orden M-QAM", [4, 16, 64], index=1)
k = int(np.log2(M))
num_simbolos = st.sidebar.slider("Símbolos", 50, 2000, 500, 50)
fc = st.sidebar.slider("Frecuencia portadora [Hz]", 10, 500, 200)
tasa_simbolo = st.sidebar.slider("Tasa de símbolo [símb/s]", 10, 1000, 100)
fs_mult = st.sidebar.slider("Factor de muestreo (fs = X * fc)", 4, 20, 10)
fs = fs_mult * fc
snr_db = st.sidebar.slider("SNR [dB]", -5, 30, 15)


# --- Simulación ---
bits_tx = generar_bits(num_simbolos, k)
simbolos_I_tx, simbolos_Q_tx, escala = mapear_bits_a_simbolos_qam(bits_tx, M, k)

signal_modulada, t = modular_qam(simbolos_I_tx, simbolos_Q_tx, fc, fs, tasa_simbolo)
signal_recibida = canal_awgn(signal_modulada, snr_db)

simbolos_I_rx, simbolos_Q_rx, bits_rx = demodular_y_decidir(signal_recibida, fc, fs, tasa_simbolo, M, k, escala)

num_errores, total_bits, ber = calcular_ber(bits_tx, bits_rx)


# --- Resultados ---
st.header(" Resultados")
st.metric("BER", f"{ber:.2e}")
st.info(f"Errores: {num_errores} / {total_bits} bits")


# --- Gráficas ---
tab1, tab2, tab3 = st.tabs([" Tiempo", " Espectro", " Constelación"])

with tab1:
    st.subheader("Señal transmitida vs recibida")
    fig, ax = plt.subplots(figsize=(12, 5))
    ax.plot(t, signal_modulada, label="Transmisión", color='dodgerblue')
    ax.plot(t, signal_recibida, label="Recepción", color='tomato', alpha=0.6)
    ax.set_xlim(0, 20 / tasa_simbolo)
    ax.set_xlabel("Tiempo [s]")
    ax.set_ylabel("Amplitud")
    ax.legend()
    st.pyplot(fig)

with tab2:
    st.subheader("Análisis espectral (FFT)")
    N_fft = len(signal_modulada)
    xf = fftshift(fftfreq(N_fft, 1 / fs))
    yf_tx = fftshift(fft(signal_modulada))
    yf_rx = fftshift(fft(signal_recibida))
    fig2, ax2 = plt.subplots(figsize=(12, 5))
    ax2.plot(xf, 20 * np.log10(np.abs(yf_tx) + 1e-12), label="Transmisión")
    ax2.plot(xf, 20 * np.log10(np.abs(yf_rx) + 1e-12), label="Recepción", alpha=0.7)
    ax2.set_xlabel("Frecuencia [Hz]")
    ax2.set_ylabel("Magnitud [dB]")
    ax2.legend()
    st.pyplot(fig2)

with tab3:
    st.subheader("Diagrama de constelación")
    fig3, ax3 = plt.subplots(figsize=(7, 7))
    ax3.scatter(simbolos_I_tx, simbolos_Q_tx, label="Ideal", marker='s', s=60, color='black')
    ax3.scatter(simbolos_I_rx, simbolos_Q_rx, label="Recibidos", color='orangered', alpha=0.5)
    ax3.axhline(0, color='gray', lw=0.8)
    ax3.axvline(0, color='gray', lw=0.8)
    ax3.set_xlabel("I")
    ax3.set_ylabel("Q")
    ax3.set_aspect('equal')
    ax3.legend()
    st.pyplot(fig3)

with st.sidebar.expander("👥 Integrantes del Proyecto"):
    st.markdown("""
    ### Equipo de Trabajo

    Este proyecto fue desarrollado por estudiantes de la **Universidad Nacional de Colombia**, como parte del curso de **de Señales y Sistemas**.

    **Integrantes:**
    - 📌 **Sofía López Usma**
    - 📌 **Miguel Andrés Peña**
    - 📌 **Juan Fernando Montoya Ortiz**


    """)
```

```python
%%writefile  E-Sistema_Completo.py
# -*- coding: utf-8 -*-


"""
Simulador de un sistema de comunicación digital con modulación QAM.

Este script crea un dashboard interactivo con Streamlit para visualizar
el efecto del ruido AWGN en una señal QAM. Permite al usuario
modificar el orden de la modulación y la relación señal a ruido (SNR)
para observar los cambios en la constelación recibida.
"""

import streamlit as st
import numpy as np
import matplotlib.pyplot as plt


custom_css = """
<style>
    .stApp {
        background-color: #0e1a40;
        color: #E0E0E0;
        font-family: 'Courier New', monospace;
    }

    h1, h2, h3 {
        color: #00BFFF;
    }

    .main .block-container {
        padding-top: 2rem;
        padding-bottom: 2rem;
        background-color: #1B1D2B;
        border-radius: 10px;
    }

    section[data-testid="stSidebar"] {
        background-color: #222f5b;
        border-radius: 10px;
    }

    section[data-testid="stSidebar"] * {
        color: white !important;
    }

    .resultado-final {
        color: #FFD700;
        background-color: #2c3e50;
        border: 1px solid #FFD700;
        border-radius: 8px;
        padding: 1rem;
        text-align: center;
        font-size: 1.1rem;
    }

    .nota-info {
        color: #98FB98;
    }
</style>
"""

st.markdown(custom_css, unsafe_allow_html=True)
# --- Configuración de la Página de Streamlit ---
st.set_page_config(
    page_title="Simulador QAM con Ruido",
    page_icon="",
    layout="wide"
)

# --- Título y Descripción ---
st.title(" Simulador de Sistema de Comunicación Digital")
st.markdown("""
Este dashboard interactivo simula un sistema de comunicación digital básico para analizar el impacto del ruido en la transmisión.

**Funcionalidades:**
1.  **Generador de Símbolos:** Crea símbolos complejos para una constelación QAM.
2.  **Canal AWGN:** Añade Ruido Blanco Gaussiano Aditivo a la señal, simulando un canal real.
3.  **Receptor:** Muestra la señal recibida en un diagrama de constelación.

Utiliza los controles en el panel de la izquierda para experimentar con diferentes configuraciones.
""")

# --- Panel Lateral de Controles ---
st.sidebar.header("Parámetros de Simulación")

qam_order = st.sidebar.selectbox(
    "Orden de Modulación (M-QAM)",
    (4, 16, 64),
    index=1,  # Valor por defecto será 16-QAM
    help="Define el número de puntos en la constelación. Un orden mayor permite transmitir más bits por símbolo."
)

num_symbols = st.sidebar.slider(
    "Número de Símbolos",
    min_value=100,
    max_value=2000,
    value=500,
    step=100,
    help="Cantidad de símbolos a transmitir en la simulación."
)

snr_db = st.sidebar.slider(
    "Relación Señal a Ruido (SNR) en dB",
    min_value=-5,
    max_value=30,
    value=15,
    step=1,
    help="Controla el nivel de ruido en el canal. Un SNR más bajo implica más ruido."
)

# --- Funciones del Sistema de Comunicación ---

def generate_qam_symbols(order, n_symbols):
    """
    Genera un array de símbolos complejos para una constelación QAM dada.

    Args:
        order (int): Orden de la modulación (4, 16, 64).
        n_symbols (int): Número de símbolos a generar.

    Returns:
        np.ndarray: Array de símbolos complejos (transmitidos).
        np.ndarray: Array de todos los puntos posibles de la constelación.
    """
    bits_per_symbol = int(np.log2(order))

    # Generar puntos de la constelación de referencia
    # Se crea una cuadrícula de puntos y se centra en el origen
    points_per_axis = int(np.sqrt(order))
    axis_values = np.arange(-(points_per_axis - 1), points_per_axis, 2)
    constellation_map_i, constellation_map_q = np.meshgrid(axis_values, axis_values)
    constellation_points = (constellation_map_i + 1j * constellation_map_q).flatten()

    # Normalizar la energía promedio de la constelación a 1
    avg_energy = np.mean(np.abs(constellation_points)**2)
    constellation_points /= np.sqrt(avg_energy)

    # Generar símbolos aleatorios seleccionando puntos de la constelación
    random_indices = np.random.randint(0, order, n_symbols)
    tx_symbols = constellation_points[random_indices]

    return tx_symbols, constellation_points

def add_awgn_noise(signal, snr_db):
    """
    Añade ruido AWGN a una señal compleja.

    Args:
        signal (np.ndarray): Señal original (compleja).
        snr_db (float): Relación señal a ruido en decibelios.

    Returns:
        np.ndarray: Señal con ruido añadido.
    """
    # La potencia de la señal es 1 gracias a la normalización
    signal_power = 1.0

    # Convertir SNR de dB a escala lineal
    snr_linear = 10**(snr_db / 10.0)

    # Calcular la potencia del ruido
    noise_power = signal_power / snr_linear

    # Generar ruido complejo (componentes I y Q son independientes)
    noise_i = np.random.normal(0, np.sqrt(noise_power / 2), len(signal))
    noise_q = np.random.normal(0, np.sqrt(noise_power / 2), len(signal))
    noise = noise_i + 1j * noise_q

    return signal + noise

def plot_constellation(tx_symbols, rx_symbols, ideal_points, snr_db, order):
    """
    Genera la figura de Matplotlib para la constelación.

    Args:
        tx_symbols (np.ndarray): Símbolos transmitidos.
        rx_symbols (np.ndarray): Símbolos recibidos (con ruido).
        ideal_points (np.ndarray): Puntos ideales de la constelación.
        snr_db (float): SNR en dB para el título del gráfico.
        order (int): Orden de QAM para el título del gráfico.

    Returns:
        matplotlib.figure.Figure: La figura generada.
    """
    fig, ax = plt.subplots(figsize=(8, 8))

    # Graficar la constelación ideal de referencia
    ax.plot(np.real(ideal_points), np.imag(ideal_points), 'kx', markersize=10, label='Puntos Ideales')

    # Graficar los símbolos recibidos
    ax.plot(np.real(rx_symbols), np.imag(rx_symbols), 'r.', alpha=0.5, label='Símbolos Recibidos')

    # Configuración del gráfico
    ax.set_title(f'Constelación Recibida para {order}-QAM (SNR = {snr_db} dB)')
    ax.set_xlabel('Componente en Fase (I)')
    ax.set_ylabel('Componente en Cuadratura (Q)')
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.axis('equal')

    # Ajustar límites para que la constelación se vea bien
    max_lim = np.max(np.abs(ideal_points)) * 1.5
    ax.set_xlim(-max_lim, max_lim)
    ax.set_ylim(-max_lim, max_lim)

    ax.legend()

    return fig

# --- Lógica Principal y Visualización ---

# 1. Generar los símbolos a transmitir
transmitted_symbols, ideal_constellation_points = generate_qam_symbols(qam_order, num_symbols)

# 2. Simular el canal añadiendo ruido
received_symbols = add_awgn_noise(transmitted_symbols, snr_db)

# 3. Mostrar la visualización
st.subheader("Diagrama de Constelación en el Receptor")
st.markdown(f"""
El siguiente gráfico muestra los símbolos tal como llegan al receptor. Los puntos ideales de la constelación de **{qam_order}-QAM** se marcan con una 'X' negra. Los puntos rojos son los símbolos recibidos después de ser afectados por el ruido del canal con un **SNR de {snr_db} dB**.
""")

# Crear y mostrar el gráfico de constelación
constellation_fig = plot_constellation(
    transmitted_symbols,
    received_symbols,
    ideal_constellation_points,
    snr_db,
    qam_order
)
st.pyplot(constellation_fig)

# --- Análisis y Conclusiones ---
st.subheader("Análisis de Resultados")
if snr_db > 20:
    st.info("Con un **SNR alto**, los puntos recibidos están muy cerca de los puntos ideales. Esto significa que un demodulador podría identificar correctamente cada símbolo con una baja probabilidad de error.")
elif snr_db < 5:
    st.warning("Con un **SNR bajo**, el ruido es muy potente. La 'nube' de puntos recibidos es tan dispersa que los símbolos se mezclan entre sí. Esto causaría una gran cantidad de errores de bit (BER alto) en la demodulación.")
else:
    st.success("Con un **SNR intermedio**, se puede apreciar claramente la dispersión causada por el ruido, pero los grupos de símbolos todavía son distinguibles. Este es un escenario típico en sistemas de comunicación reales.")

with st.sidebar.expander("👥 Integrantes del Proyecto"):
    st.markdown("""
    ### Equipo de Trabajo

    Este proyecto fue desarrollado por estudiantes de la **Universidad Nacional de Colombia**, como parte del curso de **de Señales y Sistemas**.

    **Integrantes:**
    - 📌 **Sofía López Usma**
    - 📌 **Miguel Andrés Peña**
    - 📌 **Juan Fernando Montoya Ortiz**


    """)
```
